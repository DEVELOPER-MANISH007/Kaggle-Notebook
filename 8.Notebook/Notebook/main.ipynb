###  ```3.2 — Text Cleaning Function```

````python
import re

def clean_basic_text(text):
    text = str(text)

    # Newline and tab ko space me convert
    text = re.sub(r'[\r\n\t]+', ' ', text)

    # Multiple spaces ko single space
    text = re.sub(r'\s+', ' ', text)

    # Starting / ending spaces remove
    text = text.strip()

    return text
````

````python
df["clean_text"] = df["comment_text"].apply(clean_basic_text)

df[["comment_text", "clean_text"]].head(10)
````

````python
def remove_urls(text):
    text = re.sub(
        r'(https?://\S+|www\.\S+|ftp://\S+)',
        ' ',
        text,
        flags=re.IGNORECASE
    )
    return text
````

###  ```4.6 — Handling Class Imbalance using Class Weight```

The target variable is highly imbalanced, with toxic comments forming only a small portion of the dataset.
To improve the model's ability to detect the minority class, we will use class_weight="balanced".

This automatically assigns higher weights to the minority class during training, without modifying the original TF-IDF data.

### ```Train Balanced Linear SVM```

````python
from sklearn.svm import LinearSVC

balanced_svm = LinearSVC(
    class_weight="balanced",
    random_state=42
)

balanced_svm.fit(
    X_train_tfidf,
    y_train
)
y_pred_balanced = balanced_svm.predict(X_test_tfidf)
````

````python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("=" * 60)
print("LINEAR SVM — CLASS WEIGHT BALANCED")
print("=" * 60)

print(f"Accuracy : {accuracy_score(y_test, y_pred_balanced):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_balanced):.4f}")
print(f"Recall   : {recall_score(y_test, y_pred_balanced):.4f}")
print(f"F1 Score : {f1_score(y_test, y_pred_balanced):.4f}")

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred_balanced))

print("\nClassification Report")
print(classification_report(y_test, y_pred_balanced))
````

### Observation

Using `class_weight="balanced"` significantly improved minority-class recall from 70.35% to 81.76%. However, precision decreased from 86.36% to 71.46%, resulting in a slightly lower F1 score.

Therefore, the balanced model is better for prioritizing toxic-comment detection (recall), while the original Linear SVM provides a better overall F1 score.

### ```4.7 — Hyperparameter Tuning```

### Hyperparameter Tuning

Hyperparameter tuning is used to find the best configuration of a machine learning model.

For Linear SVM, we will tune the `C` parameter.

Since the dataset is imbalanced, `F1 Score` will be used as the optimization metric instead of accuracy.

````python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import LinearSVC

svm = LinearSVC(
    random_state=42
)

param_grid = {
    "C": [0.01, 0.1, 1, 10]
}

svm_grid = GridSearchCV(
    estimator=svm,
    param_grid=param_grid,
    scoring="f1",
    cv=5,
    n_jobs=-1
)

svm_grid.fit(
    X_train_tfidf,
    y_train
)
````

````python
best_svm = svm_grid.best_estimator_

y_pred_svm = best_svm.predict(X_test_tfidf)

print("=" * 60)
print("TUNED LINEAR SVM")
print("=" * 60)

print(f"Accuracy : {accuracy_score(y_test, y_pred_svm):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_svm):.4f}")
print(f"Recall   : {recall_score(y_test, y_pred_svm):.4f}")
print(f"F1 Score : {f1_score(y_test, y_pred_svm):.4f}")

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred_svm))

print("\nClassification Report")
print(classification_report(y_test, y_pred_svm))
````

### Observation

Hyperparameter tuning of Linear SVM did not improve the test-set performance compared with the baseline Linear SVM.

The tuned model achieved an F1 score of 0.7754, which is identical to the baseline model. Therefore, the baseline Linear SVM remains the preferred model at this stage.

````python
print("Best Parameters:", svm_grid.best_params_)
print("Best CV F1:", svm_grid.best_score_)
````

### ```4.8 — Decision Threshold Tuning```

### Decision Threshold Tuning

Linear SVM normally uses a default decision boundary to classify samples.

By adjusting the decision threshold, we can control the trade-off between precision and recall.

The goal is to find a threshold that provides a better F1 Score for the toxic class.

````python
decision_scores = best_svm.decision_function(X_test_tfidf)

print("First 10 Decision Scores:")
print(decision_scores[:10])
````

````python
import numpy as np

thresholds = np.arange(-1.0, 1.01, 0.05)
````

````python
threshold_results = []

for threshold in thresholds:

    y_pred_threshold = (decision_scores >= threshold).astype(int)

    precision = precision_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })
````

````python
threshold_df = pd.DataFrame(threshold_results)

threshold_df = threshold_df.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)

threshold_df.head(10)
````

````python
## best threshold
best_threshold = threshold_df.iloc[0]["Threshold"]
best_threshold_f1 = threshold_df.iloc[0]["F1 Score"]

print("Best Threshold:", best_threshold)
print("Best F1 Score:", best_threshold_f1)
````

````python
y_pred_threshold = (
    decision_scores >= best_threshold
).astype(int)
````

````python
print("=" * 60)
print("LINEAR SVM — OPTIMIZED THRESHOLD")
print("=" * 60)

print(f"Threshold: {best_threshold:.2f}")
print(f"Accuracy : {accuracy_score(y_test, y_pred_threshold):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_threshold):.4f}")
print(f"Recall   : {recall_score(y_test, y_pred_threshold):.4f}")
print(f"F1 Score : {f1_score(y_test, y_pred_threshold):.4f}")

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred_threshold))

print("\nClassification Report")
print(classification_report(y_test, y_pred_threshold))
````

### Observation

Decision threshold tuning improved the Linear SVM's F1 score from 0.7754 to 0.7872.

The optimal threshold was -0.20, which provided a better balance between precision and recall.

Final Performance:
- Accuracy: 96.03%
- Precision: 80.87%
- Recall: 76.69%
- F1 Score: 78.72%

Therefore, the threshold-optimized Linear SVM is selected as the best model at this stage.

````python
def predict_toxicity(comment):

    # Basic cleaning
    cleaned_comment = clean_basic_text(comment)

    # URL removal
    cleaned_comment = remove_urls(cleaned_comment)

    # Lowercase
    cleaned_comment = cleaned_comment.lower()

    # Stopword removal
    cleaned_comment = remove_stopwords(cleaned_comment)

    # TF-IDF
    comment_tfidf = tfidf.transform([cleaned_comment])

    # SVM decision score
    score = best_svm.decision_function(comment_tfidf)[0]

    # Optimized threshold
    prediction = int(score >= best_threshold)

    result = "TOXIC" if prediction == 1 else "NON-TOXIC"

    print("=" * 50)
    print("Comment :", comment)
    print("Result  :", result)
    print(f"Score   : {score:.4f}")
    print(f"Threshold: {best_threshold:.2f}")
    print("=" * 50)
````
